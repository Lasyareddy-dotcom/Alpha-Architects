# EvoDrift Demo
Adaptive Multi-Objective Evolutionary Deep Learning for Robust Classification under Non-Stationary Data Distributions.

```python
from evodrift.data import make_stream
from evodrift.pipeline import run_experiment
import pandas as pd
```

```python
X, y, phase = make_stream(samples_per_phase=300, seed=42)
print('Shape:', X.shape)
print('Features: rainfall, river_level, temperature, seismic')
print('Phases:', sorted(set(phase)))
```

```python
result = run_experiment(samples_per_phase=300, seed=42, population_size=8, generations=3)
result['summary']
```

```python
result['phase_metrics']
```

```python
result['drift_events']
```

```python
plot_df = result['stream'][['index','baseline_correct','adaptive_correct']].set_index('index').rolling(30).mean()
plot_df.plot(title='30-sample rolling accuracy', figsize=(10,4))
```

## Interpretation
The baseline is frozen after warm-up. The adaptive model listens for drift and invokes the multi-objective evolutionary search only when the drift detector signals a change. The experiment reports macro-F1, accuracy, drift events, and phase-level robustness.
